### What M14.4 Will Cover

**1. Operating Model Framework**
- **Centralized Model:** Single platform team controls all operations; max consistency; suited for <10 highly regulated tenants
- **Federated Model:** Tenant teams self-manage namespaces with platform guardrails; scales to 100+ tenants
- **Hybrid Model:** Platform-owned core infrastructure + tenant champions + self-service portals (most common for GCCs with 10-100 business units)

**2. Team Sizing Calculator**
- Base ratio: 1:12 engineer-to-tenant (one engineer supports 12 tenants)
- Complexity multipliers:
  - Low complexity: 1.5× = 1:18 support level
  - Medium complexity: 1.0× = 1:12 baseline
  - High complexity: 0.75× = 1:9 support level
- Minimum: 2 engineers (redundancy/on-call)

**3. Self-Service Portal Implementation**
- Stack: React frontend + FastAPI backend + Temporal workflow engine + Open Policy Agent
- Automation target: 80% of tier 1 requests resolved without human intervention
- Capabilities: Configuration viewing, quota workflows, documentation access, real-time usage monitoring

**4. Three-Level Escalation Workflow**
- **Tier 1:** Self-Service Portal (2-min response, 80% of requests) — Documentation, automated approvals
- **Tier 2:** Tenant Champions (1-2 hours, 15% of requests) — Access grants, configuration changes
- **Tier 3:** Platform Team (variable, 5% of requests) — Platform bugs, feature development, security

**5. SLA Templates by Tenant Tier**
- **Platinum:** 99.95% uptime, 2-hour incident response, dedicated support, quarterly business reviews (₹15-20 lakh annual)
- **Gold:** 99.9% uptime, 4-hour response, shared support, monthly check-ins (₹8-12 lakh annual)
